```python
def preprocess_text(text):
    # Make text lowercase and remove links, text in square brackets, punctuation, and words containing numbers
    text = str(text)
    text = text.lower()
    text = re.sub(r'https?://\S+|www\.\S+|\[.*?\]|[^a-zA-Z\s]+|\w*\d\w*', ' ', text)

    # Remove stop words
    stop_words = set(stopwords.words("english"))
    words = text.split()
    filtered_words = [word for word in words if word not in stop_words]
    text = ' '.join(filtered_words).strip()

    # Tokenize
    tokens = nltk.word_tokenize(text)

    # Lemmatize
    lemmatizer = WordNetLemmatizer()
    lem_tokens = [lemmatizer.lemmatize(token) for token in tokens]
    
    return ' '.join(lem_tokens)
```

#### Vectorization

```python
def word_distribution(text):
    vectorizer = CountVectorizer(max_features=20000)  # limit features to reduce load
    docs = vectorizer.fit_transform(text)
    features = vectorizer.get_feature_names_out()

    # Efficient sparse sum
    counts = np.asarray(docs.sum(axis=0)).flatten()

    # Frequency dictionary
    freq_dict = dict(zip(features, counts))

    # Get top 20
    sorted_items = sorted(freq_dict.items(), key=lambda x: x[1], reverse=True)[:20]

    # Plot
    words, frequencies = zip(*sorted_items)
    plt.figure(figsize=(12, 6))
    plt.bar(words, frequencies, color='teal')
    plt.xticks(rotation=45, ha='right')
    plt.title("Top 20 Word Frequencies")
    plt.xlabel("Words")
    plt.ylabel("Frequencies")
    plt.tight_layout()
    plt.show()


word_distribution(start_data['Review'])
```

```python
tfidf_vectorizer = TfidfVectorizer(max_features = 5000, ngram_range = (1,3))
```

```python
X_tfidf = tfidf_vectorizer.fit_transform(start_data['Review'])
X_tfidf
```

```python
y = start_data["Sentiment"]
```

```python
y.shape
```

### **Training the model using Logistic Regression**

```python
X_train, X_test, y_train, y_test = train_test_split(X_tfidf, y, random_state = 42,test_size = 0.20)
```

```python
print(X_train.shape)
print(f"Number of Training Samples: {X_train.shape[0]}")
print(y_train.shape)
print(f"Number of Training Samples: {X_train.shape[0]}")
print(X_test.shape)
print(f"Number of Training Samples: {X_train.shape[0]}")
print(y_test.shape)
print(f"Number of Training Samples: {X_train.shape[0]}")
```

```python
label_encoder = LabelEncoder()
y_train_encoded = label_encoder.fit_transform(y_train)
y_test_encoded = label_encoder.transform(y_test)
```

```python

# Logistic Regression
lr = LogisticRegression(max_iter=1000)
lr.fit(X_train, y_train_encoded)
lr_preds = lr.predict(X_test)
lr_acc = accuracy_score(y_test_encoded, lr_preds)
print(f"Logistic Regression: Accuracy = {lr_acc:.4f}")

```

```python


# Random Forest
rf = RandomForestClassifier(n_estimators=100)
rf.fit(X_train, y_train_encoded)
rf_preds = rf.predict(X_test)
rf_acc = accuracy_score(y_test_encoded, rf_preds)
print(f"Random Forest: Accuracy = {rf_acc:.4f}")
```

```python

# SVM
svm = SVC()
svm.fit(X_train, y_train_encoded)
svm_preds = svm.predict(X_test)
svm_acc = accuracy_score(y_test_encoded, svm_preds)
print(f"SVM: Accuracy = {svm_acc:.4f}")
```

```python

# Naive Bayes
nb = MultinomialNB()
nb.fit(X_train, y_train_encoded)
nb_preds = nb.predict(X_test)
nb_acc = accuracy_score(y_test_encoded, nb_preds)
print(f"Naive Bayes: Accuracy = {nb_acc:.4f}")
```

```python

# Decision Tree
dt = DecisionTreeClassifier()
dt.fit(X_train, y_train_encoded)
dt_preds = dt.predict(X_test)
dt_acc = accuracy_score(y_test_encoded, dt_preds)
print(f"Decision Tree: Accuracy = {dt_acc:.4f}")
```

```python

# K-Nearest Neighbors
knn = KNeighborsClassifier()
knn.fit(X_train, y_train_encoded)
knn_preds = knn.predict(X_test)
knn_acc = accuracy_score(y_test_encoded, knn_preds)
print(f"KNN: Accuracy = {knn_acc:.4f}")
```

```python
logreg = LogisticRegression()
logreg.fit(X_train, y_train)
```

```python
predictedLogReg = logreg.predict(X_test)
np.mean(predictedLogReg == y_test)
```

```python
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy * 100:.2f}%")
```

```python
plt.figure(figsize=(5,5))
sns.heatmap(confusion_matrix(y_test,predictedLogReg), annot=True, fmt="d", cmap=colors)
plt.show()
```

```python
cm = confusion_matrix(y_test,predictedLogReg)

print('Confusion matrix\n\n', cm)
print('\nTrue Positives(TP) = ', cm[0,0])
print('\nTrue Negatives(TN) = ', cm[1,1])
print('\nFalse Positives(FP) = ', cm[0,1])
print('\nFalse Negatives(FN) = ', cm[1,0])
```

```python
TP = cm[0,0]
TN = cm[1,1]
FP = cm[0,1]
FN = cm[1,0]
```

```python
classification_accuracy = (TP + TN) / float(TP + TN + FP + FN)
print('Classification accuracy of LG: {0:0.4f}'.format(classification_accuracy))
```

```python
classification_error = (FP + FN) / float(TP + TN + FP + FN)
print('Classification error of LG: {0:0.4f}'.format(classification_error))
```

```python
precision = TP / float(TP + FP)
print('Precision of LG: {0:0.4f}'.format(precision))
```

```python
recall = TP / float(TP + FN)
print('Recall or Sensitivity of LG: {0:0.4f}'.format(recall))
```
